```python
print("----- STUDENT GRADE CALCULATOR -----")
student_name=input("Enter student name: ")
marks = int(input("Enter Marks(0-100): "))
if marks < 0 or marks > 100:
    print("INVALID MARKS! Please enter marks between 0 and 100.")
elif marks >= 90:
    grade = "A+"
    Remark = "Excellent"
    Result = "PASS!"
elif marks >=80:
    grade = "A"
    Remark = "Very Good"
    Result = "PASS!"
elif marks >=70:
    grade = "B"
    Remark = "Good"
    Result = "PASS!"
elif marks >=60:
    grade = "C"
    Remark = "Satisfactory"
    Result = "PASS!"
elif marks >=50:
    grade = "D"
    Remark = "Average"
    Result = "PASS!"
elif marks >=35:
    grade = "E"
    Remark = "Work Hard"
    Result = "FAIL!"
else:
    grade = "F"
    Remark = "Poor"
    Result = "FAIL!"

if 0 <= marks <= 100:
    print("\n----- STUDENT GRADE REPORT -----")
    print("Student Name:", student_name)
    print("Marks:", marks)
    print("Grade:", grade)
    print("Remark:", Remark)
    print("Result:", Result)

```
